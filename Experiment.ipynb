## Performance Improvement

- Achieves **a lower miss ratio** than CLOCK across both low and high cache sizes, making it a more efficient replacement policy.
- Compared to **ARC (Adaptive Replacement Cache)**, CAR exhibits a lower miss ratio at higher cache sizes.

The size of the cache can be defined as:

- Size (e.g. 1mb, 16mb, 8gb)
- Fraction of the working set size (e.g. 0.001, 0.1, 0.2)
- Number of object if the flag --ignore-obj-size 1 is enabled (e.g. 1000, 16000)

Following is the general format for running CAR:
```
cachesim <trace path> <trace format> clockpro <cache size>
```

Example:
```
cachesim data/trace.vscsi vscsi car 0.001,0.01,0.1,0.2 -e init-ref=0,init-ratio-cold=1
```
